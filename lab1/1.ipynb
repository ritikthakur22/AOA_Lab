## 3. Related Theory

Iterative and numeric algorithms form the backbone of computer science problem-solving.

### Iterative Algorithms
Iterative algorithms use repetitive control structures (like `for` and `while` loops) to execute a set of instructions until a specific condition is met. 
- **Searching**: Sequential Search is the most basic iterative search, iterating through elements one by one until the target is found. It's unoptimized but works on unsorted data.
- **Sorting**: Algorithms like Bubble Sort, Insertion Sort, and Selection Sort are iterative comparison-based sorting mechanisms. They iteratively compare and swap elements to build a sorted sequence.

### Numeric & Number Theory Algorithms
These algorithms perform mathematical calculations, often on integers, and are foundational to modern cryptography.
- **Euclidean Algorithm (GCD)**: An efficient method for computing the greatest common divisor of two integers. It uses the principle that the GCD of two numbers also divides their difference.
- **Fibonacci Series**: A sequence where each number is the sum of the two preceding ones. Iterative implementation is preferred over recursion to avoid exponential time complexity.
- **Chinese Remainder Theorem (CRT)**: A theorem of number theory stating that if one knows the remainders of the Euclidean division of an integer n by several integers, then one can determine uniquely the remainder of the division of n by the product of these integers, under the condition that the divisors are pairwise coprime.
- **Fermat's Theorem (Primality Test)**: Based on Fermat's Little Theorem. It is a probabilistic test to determine whether a number is a probable prime.
- **Miller-Rabin Test**: A more advanced probabilistic primality test that overcomes the limitations of Fermat's test (such as Carmichael numbers) by checking for non-trivial square roots of 1 modulo n.

### 5.1. GCD using the Euclid algorithm

**Algorithm:**
1. Start.
2. Input two integers `a` and `b`.
3. While `b` is not equal to 0, do the following:
   a. Calculate the remainder `r = a % b`.
   b. Assign `a = b`.
   c. Assign `b = r`.
4. Return `a` as the GCD.
5. Stop.


````python
def gcd_euclid(a, b):
    print(f"Finding GCD for {a} and {b}")
    while b != 0:
        print(f" -> a = {a}, b = {b}, remainder = {a % b}")
        a, b = b, a % b
    return a

num1, num2 = 48, 18
result = gcd_euclid(num1, num2)
print(f"\nFinal GCD of {num1} and {num2} is: {result}")
````

### 5.2. Fibonacci Series

**Algorithm:**
1. Start.
2. Input the number of terms `n`.
3. If `n <= 0`, return an empty list.
4. If `n == 1`, return `[0]`.
5. Initialize an array `fib_series = [0, 1]`.
6. For `i` from 2 to `n-1`:
   a. Append `(fib_series[-1] + fib_series[-2])` to `fib_series`.
7. Return `fib_series`.
8. Stop.


````python
def fibonacci(n):
    if n <= 0: return []
    if n == 1: return [0]
    
    fib_series = [0, 1]
    for i in range(2, n):
        fib_series.append(fib_series[-1] + fib_series[-2])
    return fib_series

n_terms = 15
print(f"First {n_terms} terms of Fibonacci series:\n{fibonacci(n_terms)}")
````

### 5.3. Sequential Search

**Algorithm:**
1. Start.
2. Input array `arr` and search `target`.
3. Loop `i` from 0 to length of `arr` - 1:
   a. If `arr[i] == target`, print success and return `i`.
4. If loop finishes without returning, print not found and return -1.
5. Stop.


````python
def sequential_search(arr, target):
    print(f"Searching for {target} in {arr}...")
    for i in range(len(arr)):
        if arr[i] == target:
            print(f" -> Found at index {i} after {i+1} comparisons.")
            return i
    print(" -> Not found.")
    return -1

arr = [10, 23, 45, 70, 11, 15]
index = sequential_search(arr, 70)
````

### 5.4. Bubble Sort

**Algorithm:**
1. Start.
2. Input array `arr` of size `n`.
3. Loop `i` from 0 to `n-1`:
   a. Initialize `swapped = False`.
   b. Loop `j` from 0 to `n-i-2`:
      i. If `arr[j] > arr[j+1]`, swap them and set `swapped = True`.
   c. If `swapped` is `False`, break loop (array is sorted).
4. Return sorted `arr`.
5. Stop.


````python
def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        swapped = False
        # Last i elements are already in place
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        # If no two elements were swapped by inner loop, then break
        if not swapped:
            break
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Original array: {arr}")
print(f"Sorted array:   {bubble_sort(arr.copy())}")
````

### 5.5. Insertion Sort

**Algorithm:**
1. Start.
2. Input array `arr` of size `n`.
3. Loop `i` from 1 to `n-1`:
   a. Set `key = arr[i]` and `j = i - 1`.
   b. While `j >= 0` and `key < arr[j]`:
      i. Move `arr[j]` to `arr[j+1]`.
      ii. Decrement `j` by 1.
   c. Insert `key` at `arr[j+1]`.
4. Return sorted `arr`.
5. Stop.


````python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # Move elements of arr[0..i-1], that are greater than key,
        # to one position ahead of their current position
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

arr = [12, 11, 13, 5, 6]
print(f"Original array: {arr}")
print(f"Sorted array:   {insertion_sort(arr.copy())}")
````

### 5.6. Selection Sort

**Algorithm:**
1. Start.
2. Input array `arr` of size `n`.
3. Loop `i` from 0 to `n-1`:
   a. Set `min_idx = i`.
   b. Loop `j` from `i+1` to `n-1`:
      i. If `arr[j] < arr[min_idx]`, update `min_idx = j`.
   c. Swap `arr[i]` and `arr[min_idx]`.
4. Return sorted `arr`.
5. Stop.


````python
def selection_sort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[min_idx] > arr[j]:
                min_idx = j
        # Swap the found minimum element with the first element
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

arr = [64, 25, 12, 22, 11]
print(f"Original array: {arr}")
print(f"Sorted array:   {selection_sort(arr.copy())}")
````

### 5.7. Chinese Remainder Theorem

**Algorithm:**
1. Start.
2. Input arrays `num` (moduli) and `rem` (remainders).
3. Calculate the total product of all elements in `num` (let's call it `prod`).
4. Initialize `result = 0`.
5. For each `i` from 0 to length of `num`:
   a. Calculate partial product `pp = prod / num[i]`.
   b. Find modular inverse `inv` of `pp` under modulo `num[i]` using Extended Euclidean Algorithm.
   c. Add `rem[i] * inv * pp` to `result`.
6. Return `result % prod`.
7. Stop.


````python
def extended_gcd(a, b):
    if a == 0:
        return b, 0, 1
    gcd, x1, y1 = extended_gcd(b % a, a)
    x = y1 - (b // a) * x1
    y = x1
    return gcd, x, y

def mod_inverse(a, m):
    gcd, x, y = extended_gcd(a, m)
    if gcd != 1:
        raise Exception("Modular inverse does not exist")
    return x % m

def chinese_remainder_theorem(num, rem):
    prod = 1
    for n in num:
        prod *= n
        
    result = 0
    for i in range(len(num)):
        pp = prod // num[i]
        result += rem[i] * mod_inverse(pp, num[i]) * pp
        
    return result % prod

num = [3, 5, 7]
rem = [2, 3, 2]
ans = chinese_remainder_theorem(num, rem)
print(f"Equations:\n x ≡ 2 (mod 3)\n x ≡ 3 (mod 5)\n x ≡ 2 (mod 7)")
print(f"\nSolution: x = {ans} (mod {3*5*7})")
````

### 5.8. Fermat's Theorem (Primality Test)

**Algorithm:**
1. Start.
2. Input integer `n` (number to test) and `k` (number of iterations).
3. If `n <= 1` or `n == 4`, return False.
4. If `n <= 3`, return True.
5. Loop `k` times:
   a. Choose a random integer `a` in range `[2, n-2]`.
   b. If `(a^(n-1)) % n != 1`, return False (Composite).
6. Return True (Probable Prime).
7. Stop.


````python
import random

def power(a, n, p):
    res = 1
    a = a % p
    while n > 0:
        if n % 2 == 1:
            res = (res * a) % p
        n = n // 2
        a = (a * a) % p
    return res

def fermat_test(n, k=5):
    if n <= 1 or n == 4: return False
    if n <= 3: return True
        
    for _ in range(k):
        a = random.randint(2, n - 2)
        if power(a, n - 1, n) != 1:
            return False
    return True

test_nums = [17, 21, 97, 561] # 561 is a Carmichael number
print("Fermat Primality Test Results:")
for n in test_nums:
    res = "Prime (Probable)" if fermat_test(n) else "Composite"
    print(f" - {n:>3} is {res}")
````

### 5.9. Miller-Rabin Test

**Algorithm:**
1. Start.
2. Input integer `n` and `k` (number of iterations).
3. If `n <= 1` or `n == 4`, return False. If `n <= 3`, return True.
4. Find `d` such that `n-1 = d * 2^r` (keep dividing `d` by 2 until it's odd).
5. Loop `k` times:
   a. Pick random `a` in range `[2, n-2]`.
   b. Compute `x = (a^d) % n`.
   c. If `x == 1` or `x == n-1`, continue to next iteration.
   d. While `d != n-1`:
      i. Compute `x = (x*x) % n`.
      ii. Multiply `d` by 2.
      iii. If `x == 1`, return False.
      iv. If `x == n-1`, break and continue to next iteration of `k`.
   e. If loop `d != n-1` finishes without finding `x == n-1`, return False.
6. Return True.
7. Stop.


````python
import random

def miller_rabin_test(d, n):
    a = random.randint(2, n - 2)
    x = power(a, d, n) # Using power from Fermat
    
    if x == 1 or x == n - 1:
        return True
        
    while d != n - 1:
        x = (x * x) % n
        d *= 2
        
        if x == 1:
            return False
        if x == n - 1:
            return True
    return False

def is_prime_mr(n, k=5):
    if n <= 1 or n == 4: return False
    if n <= 3: return True
        
    d = n - 1
    while d % 2 == 0:
        d //= 2
        
    for _ in range(k):
        if not miller_rabin_test(d, n):
            return False
    return True

print("Miller-Rabin Primality Test Results:")
for n in test_nums:
    res = "Prime (Probable)" if is_prime_mr(n) else "Composite"
    print(f" - {n:>3} is {res}")
````
