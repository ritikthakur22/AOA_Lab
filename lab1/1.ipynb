## Related Theory
**Iterative Algorithms**: These algorithms use repetitive constructs like loops to solve a problem step-by-step. Examples include Bubble Sort, Insertion Sort, and Sequential Search.

**Numeric/Number Theory Algorithms**: These deal with mathematical computations on integers. Examples include Euclid's Algorithm for GCD, Chinese Remainder Theorem, Fermat's Primality Test, and the Miller-Rabin Test.

## Computer Code & Outputs

### 1. GCD using the Euclid algorithm

```python
def gcd_euclid(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# Test the algorithm
num1, num2 = 48, 18
print(f"GCD of {num1} and {num2} is {gcd_euclid(num1, num2)}")
```

### 2. Fibonacci Series

```python
def fibonacci(n):
    fib_series = []
    a, b = 0, 1
    for _ in range(n):
        fib_series.append(a)
        a, b = b, a + b
    return fib_series

# Test the algorithm
n_terms = 10
print(f"First {n_terms} terms of Fibonacci series: {fibonacci(n_terms)}")
```

### 3. Sequential Search

```python
def sequential_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Test the algorithm
arr = [10, 23, 45, 70, 11, 15]
target = 70
index = sequential_search(arr, target)
print(f"Element {target} found at index: {index}")
```

### 4. Bubble Sort

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

# Test the algorithm
arr = [64, 34, 25, 12, 22, 11, 90]
sorted_arr = bubble_sort(arr.copy())
print(f"Sorted array: {sorted_arr}")
```

### 5. Insertion Sort

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Test the algorithm
arr = [12, 11, 13, 5, 6]
sorted_arr = insertion_sort(arr.copy())
print(f"Sorted array: {sorted_arr}")
```

### 6. Selection Sort

```python
def selection_sort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[min_idx] > arr[j]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# Test the algorithm
arr = [64, 25, 12, 22, 11]
sorted_arr = selection_sort(arr.copy())
print(f"Sorted array: {sorted_arr}")
```

### 7. Chinese Remainder Theorem

```python
def extended_gcd(a, b):
    if a == 0:
        return b, 0, 1
    gcd, x1, y1 = extended_gcd(b % a, a)
    x = y1 - (b // a) * x1
    y = x1
    return gcd, x, y

def mod_inverse(a, m):
    gcd, x, y = extended_gcd(a, m)
    if gcd != 1:
        raise Exception("Modular inverse does not exist")
    else:
        return x % m

def chinese_remainder_theorem(num, rem):
    prod = 1
    for n in num:
        prod *= n
        
    result = 0
    for i in range(len(num)):
        pp = prod // num[i]
        result += rem[i] * mod_inverse(pp, num[i]) * pp
        
    return result % prod

# Test the algorithm
# x = 2 (mod 3), x = 3 (mod 5), x = 2 (mod 7)
num = [3, 5, 7]
rem = [2, 3, 2]
print(f"x = {chinese_remainder_theorem(num, rem)} (mod {3*5*7})")
```

### 8. Fermat's Theorem (Primality Test)

```python
import random

def power(a, n, p):
    res = 1
    a = a % p
    while n > 0:
        if n % 2 == 1:
            res = (res * a) % p
        n = n // 2
        a = (a * a) % p
    return res

def fermat_test(n, k=5):
    if n <= 1 or n == 4:
        return False
    if n <= 3:
        return True
        
    for _ in range(k):
        a = random.randint(2, n - 2)
        if power(a, n - 1, n) != 1:
            return False
            
    return True

# Test the algorithm
test_nums = [17, 21, 97, 100]
for n in test_nums:
    res = "Prime" if fermat_test(n) else "Composite"
    print(f"{n} is {res}")
```

### 9. Miller-Rabin Test

```python
import random

# We use the power function from Fermat's Test block
def miller_rabin_test(d, n):
    a = random.randint(2, n - 2)
    x = power(a, d, n)
    
    if x == 1 or x == n - 1:
        return True
        
    while d != n - 1:
        x = (x * x) % n
        d *= 2
        
        if x == 1:
            return False
        if x == n - 1:
            return True
            
    return False

def is_prime(n, k=5):
    if n <= 1 or n == 4:
        return False
    if n <= 3:
        return True
        
    d = n - 1
    while d % 2 == 0:
        d //= 2
        
    for _ in range(k):
        if not miller_rabin_test(d, n):
            return False
            
    return True

# Test the algorithm
test_nums = [17, 21, 97, 100]
for n in test_nums:
    res = "Prime" if is_prime(n) else "Composite"
    print(f"{n} is {res}")
```

## Analysis of the algorithms
- **GCD (Euclid)**: Time Complexity $O(\log(\min(a, b)))$
- **Fibonacci**: Time Complexity $O(n)$ for generation.
- **Sequential Search**: Time Complexity $O(n)$
- **Bubble, Insertion, Selection Sort**: Time Complexity $O(n^2)$
- **CRT**: Time Complexity $O(k \log(\prod m_i))$
- **Fermat & Miller-Rabin**: Time Complexity $O(k \log^3 n)$
